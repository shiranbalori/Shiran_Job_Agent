```python
def match_candidate_to_job(profile, job):

    score = 0

    matched_skills = []

    for skill in profile.skills:

        if skill.lower() in job.description.lower():

            score += 1

            matched_skills.append(skill)

    return {
        "score": score,
        "matched_skills": matched_skills
    }
```

```python
result = match_candidate_to_job(profile, job)
```

```python
print(result["score"])
```

```python
print(result["matched_skills"])
```

```python
import requests
import pandas as pd

companies = [
    "riskified",
    "similarweb",
    "taboola",
    "appsFlyer",
    "monday"
]

all_jobs = []

for company in companies:
    url = f"https://api.greenhouse.io/v1/boards/{company}/jobs?content=true"
    response = requests.get(url)

    print(company, response.status_code)

    try:
        data = response.json()
    except:
        print("Not JSON")
        continue

    if "jobs" not in data:
        print("No jobs:", data)
        continue

    for job in data["jobs"]:
        location = job.get("location", {}).get("name", "")
        content = job.get("content", "")

        location_text = location.lower()

        if (
            "israel" in location_text
            or "tel aviv" in location_text
            or "herzliya" in location_text
            or "ramat gan" in location_text
        ):
            all_jobs.append({
                "title": job.get("title"),
                "company": company,
                "location": location,
                "description": content,
                "link": job.get("absolute_url")
            })

jobs_df = pd.DataFrame(all_jobs)

jobs_df
```

```python
def create_cover_letter(row):
    return f"""
Dear Hiring Team,

I am excited to apply for the {row['title']} role at {row['company']}.

My background combines AI automation, Python, LLM-based workflows, RAG systems,
document understanding, APIs, and business-facing AI solutions. I have hands-on
experience building practical AI workflows that connect technical capabilities
with real business needs.

This role caught my attention because it appears to involve building practical
technology solutions and working across business and technical requirements.
I would be happy to bring my experience in AI-driven automation, enterprise
workflows, and applied AI solutions to your team.

Best regards,
Shiran Balori
"""
```

```python
results_df = jobs_df.copy()

results_df["cover_letter"] = results_df.apply(create_cover_letter, axis=1)

results_df[["title", "company", "location", "cover_letter"]]
```
